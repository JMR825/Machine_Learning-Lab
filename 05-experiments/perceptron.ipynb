```python
import torch
import torch.nn as nn
import torch.optim as optim

class SimpleNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.layer1=nn.Linear(13, 16)
        self.layer2=nn.Linear(16,3)
        
    def forward(self,x):
        x=self.layer1(x)
        x=self.layer2(x)
        return x
```
